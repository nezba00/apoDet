```python
import numpy as np
import tifffile
import matplotlib.pyplot as plt
from ipywidgets import interact


def visualize_frames(frame_index):
    # Create a figure with two subplots.
    fig, axs = plt.subplots(1, 2, figsize=(12, 6))
    
    # Left subplot: display timelapse1 with axes and grid.
    im1 = axs[0].imshow(timelapse1[frame_index], cmap='gray')
    axs[0].set_title(f"Timelapse 1 - Frame {frame_index}")
    axs[0].set_xlabel("X (pixels)")
    axs[0].set_ylabel("Y (pixels)")
    axs[0].set_xticks(np.arange(0, timelapse1.shape[2], 100))
    axs[0].set_yticks(np.arange(0, timelapse1.shape[1], 100))
    axs[0].grid(True, which='both', linestyle='--', linewidth=0.5)
    
    # Right subplot: overlay timelapse2 binary mask over timelapse1.
    # First, display timelapse1 as the base image.
    axs[1].imshow(timelapse1[frame_index], cmap='gray')
    # Then, create a masked array from timelapse2 so that zeros are transparent.
    mask = np.ma.masked_where(timelapse2[frame_index] == 0, timelapse2[frame_index])
    # Overlay the mask. The 'jet' cmap is used to highlight the mask overlay, adjust as needed.
    axs[1].imshow(mask, cmap='twilight', alpha=0.5)
    
    axs[1].set_title(f"Overlay of Timelapse 2 on Timelapse 1 - Frame {frame_index}")
    axs[1].set_xlabel("X (pixels)")
    axs[1].set_ylabel("Y (pixels)")
    axs[1].set_xticks(np.arange(0, timelapse1.shape[2], 100))
    axs[1].set_yticks(np.arange(0, timelapse1.shape[1], 100))
    axs[1].grid(True, which='both', linestyle='--', linewidth=0.5)
    
    plt.tight_layout()
    plt.show()

# Create an interactive slider to navigate through frames.
interact(visualize_frames, frame_index=(0, timelapse1.shape[0]-1))
```
